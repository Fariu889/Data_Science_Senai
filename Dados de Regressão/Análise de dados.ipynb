```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tabela =  pd.read_excel("dados_regressao_linear.xlsx")

tabela
```

```python
media_x = tabela["Investimento_X"].mean() # cálculo da média do x
media_y = tabela["Vendas_Y"].mean()# cálculo da média do y

tabela["dist_x"] = tabela["Investimento_X"] - media_x # o cálculo da distancia de x até a média de x
tabela["dist_y"] = tabela["Vendas_Y"] - media_y# o cálculo da distancia de y até a média de y

tabela["dist_x_quadrado"] = tabela["dist_x"]**2# cálculo do quadrado da distancia de x

a = (np.sum(tabela["dist_x"]* tabela["dist_y"])# Distancia dos pontos
     / # divisão
     np.sum(tabela["dist_x_quadrado"]))# do quadrado de X para descobrir o a
b = media_y - (a * media_x) # cálculo para descobri o b

prox_invest = 10

previsao = a * prox_invest + b # Formula da previsão

print(previsao)
```

```python
sns.scatterplot(data = tabela, x = "Investimento_X", y= "Vendas_Y", hue = "Ponto", palette = "muted")
sns.lineplot(data = tabela, x = "Investimento_X", y= "Vendas_Y", color = "gray")
```
