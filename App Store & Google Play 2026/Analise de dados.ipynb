```python
conversao = {
    "USD": 1.0,
    "BRL": 0.18,
    "EUR": 1.08,
    "GBP": 1.27,
    "JPY": 0.0065,
    "CAD": 0.72,
    "AUD": 0.65,
    "INR": 0.012,
    "KRW": 0.00072,
    "MXN": 0.050
}


df["price_USD"] = df["price"] * df["currency"].map(conversao)

df[df["price_USD"]> 0]
```

```python
df["primary_genre"] = df["genres"].str.split(";").str[0]

df
```
