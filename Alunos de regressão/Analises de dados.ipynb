```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tabela = pd.read_excel("atividade_regressao_linear_alunos.xlsx")

tabela

```

```python
media_x = tabela["Horas de Estudo (X)"].mean() # cálculo da média do x
media_y = tabela["Nota na Prova (Y)"].mean()# cálculo da média do y

tabela["dist_x"] = tabela["Horas de Estudo (X)"] - media_x # o cálculo da distancia de x até a média de x
tabela["dist_y"] = tabela["Nota na Prova (Y)"] - media_y# o cálculo da distancia de y até a média de y

tabela["dist_x_quadrado"] = tabela["dist_x"]**2# cálculo do quadrado da distancia de x

a = (np.sum(tabela["dist_x"]* tabela["dist_y"])# Distancia dos pontos
     / # divisão
     np.sum(tabela["dist_x_quadrado"]))# do quadrado de X para descobrir o a
b = media_y - (a * media_x) # cálculo para descobri o b

prox_invest = 100

previsao = (prox_invest - b) / a # Formula da previsão

print(previsao)
```
