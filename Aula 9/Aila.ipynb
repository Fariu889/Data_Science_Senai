```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

investimentos = pd.read_excel("Datasets/dados_regressao_linear.xlsx")
```

```python
sns.scatterplot(
    data = investimentos,
    y = "Vendas_Y",
    x = "Investimento_X"
)
plt.show
```

```python
media_x = investimentos["Investimento_X"].mean() # Calcular a média de X
media_y = investimentos["Vendas_Y"].mean() # Calcular a média de Y

# (x - media_x) * (y - media_y) - Nessa parte estamos fazendo apenas a subtração a multiplicação está sendo feita mais abaixo
investimentos["dist_x"] = investimentos["Investimento_X"] - media_x # Distancia de X até a média de X
investimentos["dist_y"] = investimentos["Vendas_Y"] - media_y # Distancia de Y até a média de Y

# (x-3)²
investimentos["dist_x_quadrado"] = investimentos["dist_x"] ** 2 # Calcular o quadrado da distancia de X

a = (
    # A continuação da fórmula (x - media_x) * (y - media_y) - mas agora apenas a multiplicação
    np.sum(investimentos["dist_x"] * investimentos["dist_y"]) # Distância dos pontos
    # a = (x - media_x) * (y - media_y)/(x-3)²
    / np.sum(investimentos["dist_x_quadrado"]) # Divisão pelo quadrado de X para descobrir o a
)
# Y = aX + b
b = media_y - (a * media_x) # Calculo para descobri o b

prox_invest = 10
# Y = aX + b
previsao = a * prox_invest + b # Fórmula de previsão

print(previsao)
```

```python
alunos = pd.read_excel("Datasets/atividade_regressao_linear_alunos.xlsx")
media_x = alunos["Horas de Estudo (X)"].mean()
media_y = alunos["Nota na Prova (Y)"].mean()

alunos["dist_x"] = alunos["Horas de Estudo (X)"] - media_x
alunos["dist_y"] = alunos["Nota na Prova (Y)"] - media_y

alunos["dist_x_quadrado"] = alunos["dist_x"] ** 2

soma_dist = np.sum(alunos["dist_x"] * alunos["dist_y"])
soma_quadrado = np.sum(alunos["dist_x_quadrado"])

a = soma_dist / soma_quadrado
b = media_y - (a * media_x)

nota_desejada = 100
horas = (nota_desejada - b) / a

print("a =", a)
print("b =", b)
print("Horas necessárias para tirar 100:", horas)
```
